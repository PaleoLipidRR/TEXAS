```python
import plotly.graph_objects as go

# 1) prepare your data
x = [1,2,3,4,5]
y1 = [2,1,3,5,4]
y2 = [1,3,2,4,6]

# 2) make traces
trace1 = go.Scatter(x=x, y=y1, name="Series 1")
trace2 = go.Scatter(x=x, y=y2, name="Series 2")

# 3) build the figure with both traces
fig = go.Figure(data=[trace1, trace2])

# 4) add a dropdown menu
fig.update_layout(
    updatemenus=[{
        "buttons": [
            {
                "label": "Show Both",
                "method": "update",
                "args": [{"visible": [True, True]}]
            },
            {
                "label": "Only Series 1",
                "method": "update",
                "args": [{"visible": [True, False]}]
            },
            {
                "label": "Only Series 2",
                "method": "update",
                "args": [{"visible": [False, True]}]
            },
        ],
        "direction": "down",
        "showactive": True,
        "x": 0.1,    # dropdown position
        "y": 1.1
    }]
)

# Set fixed x-axis and y-axis limits
fig.update_layout(
    xaxis=dict(range=[0, 6]),  # x-axis limits
    yaxis=dict(range=[0, 7])   # y-axis limits
)

# Keep the legend in the same position at the right of the plot even if the plot selects only one series
fig.update_layout(
    legend=dict(
        x=1.05,  # x position of the legend
        y=1,     # y position of the legend
        traceorder="normal",
        orientation="v"
    )
)

display(fig)

# 5) export to a single HTML file
# fig.write_html("my_plot.html", include_plotlyjs='cdn', full_html=True)
```
