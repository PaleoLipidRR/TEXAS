```python
import psutil
import os

# System memory
mem = psutil.virtual_memory()
print(f"Total: {mem.total / 1024**3:.2f} GB")
print(f"Available: {mem.available / 1024**3:.2f} GB")
print(f"Used: {mem.used / 1024**3:.2f} GB ({mem.percent}%)")
print(f"Free: {mem.free / 1024**3:.2f} GB")

# Current process
process = psutil.Process(os.getpid())
print(f"\nCurrent Python process: {process.memory_info().rss / 1024**3:.2f} GB")

# Check Docker memory limit
try:
    with open('/sys/fs/cgroup/memory/memory.limit_in_bytes', 'r') as f:
        limit = int(f.read().strip())
        if limit < 9223372036854771712:  # Not unlimited
            print(f"Docker limit: {limit / 1024**3:.2f} GB")
        else:
            print("Docker limit: unlimited")
except:
    print("Cannot read memory limit")
```
