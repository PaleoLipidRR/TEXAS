```python
from pathlib import Path
import re
import os

fpath = Path(f"/home/micromamba/app/notebooks")
fname = "references.bib"

bib_file = fpath / fname

content = bib_file.read_text()
cleaned = re.sub(r'\s*(url|mendeley-groups|mendeley-tags|file|abstract)\s*=\s*\{[^{}]*\},?\n?', '\n', content)

output_file = fpath / "references_cleaned.bib"
output_file.write_text(cleaned)
```
